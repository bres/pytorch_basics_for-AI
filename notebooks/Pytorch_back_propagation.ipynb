```python
import torch
```

```python
x =torch.tensor(1.0)
y =torch.tensor(2.0)
# Initial weight
w = torch.tensor(1.0,requires_grad=True)
```

```python
# The forwrad pass to compute the loss
y_hat = w*x
loss =(y_hat-y)**2
print(loss)
```

```python
# The backward pass - pytorch will compute the local gradients automatically
# and also computes the backward pass
loss.backward()
print(w.grad)
```
