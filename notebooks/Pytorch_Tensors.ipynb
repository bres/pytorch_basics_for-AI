```python
## Broadcasting
## Allows you to perform operations on tensors with different shapes as long as they are compatible
tensor_c = torch.tensor([1,2,3])
tensor_d = torch.tensor([[1],[2],[3]])
broadcast_result = tensor_c + tensor_d
broadcast_result
```

```python
## Extract the second row
second_row = tensor_a[1,:]
second_row
```

```python
## Extract the third col
third_col = tensor_a[:,2]
third_col
```

# **Automatic Differentiation**
## Gradients are crusial for updating model parameters during training
## In pytorch

```python
## Autograd automates the process of calculating gradients, in pytorch every tensor has a flag, requires_grad, if requires_grad is True pytorch will track all operation in the tensor and automatically compute gradients during back propagation.

x = torch.tensor([2.0,3.0],requires_grad=True)
x
```

```python
y = x*3
```

```python
z = y.sum()
```

```python
y
```

```python
z
```

```python
z.backward()
```

```python
x.grad
```

```python
## No gradients
with torch.no_grad():
  y = x*2
  print("y without gradient tracking:",y)
```

```python
## Resseting gradients to zero
x.grad.zero_()
```

```python
z =(x ** 2).sum()
```

```python
z
```

```python
z.backward()
```

```python
## z = x[0]^2 +x[1]^2 w.r.t x -->2*x[0]+2*x[1]=2*2 , 2*3=4, 6
x.grad
```
